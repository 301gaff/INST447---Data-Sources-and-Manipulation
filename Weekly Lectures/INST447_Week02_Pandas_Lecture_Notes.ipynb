## DataFrame fundamentals

Each row represents one flight; each column represents a field. A DataFrame is a labeled two-dimensional table whose columns can have different types. A Series is one-dimensional. A numerical matrix usually uses a common numeric type; higher-dimensional arrays are a useful mental model for tensors.

The original ten flight records intentionally contain string distances, empty strings, missing prices, and missing delays. The helper below prints all example outputs without needing IPython.

````python
import pandas as pd

def display(*objects):
    for obj in objects:
        print(obj)
        print()

flights_data = [
    ("2024-01-15", "UA1247", "BWI", "ORD", "651", "B737", "12A", 289.50, 15),
    ("2024-01-22", "DL456", "ORD", "LAX", "1745", "A321", "8F", 425.00, None),
    ("2024-02-08", "WN2891", "LAX", "PHX", "370", "B737", "", 149.99, 0),
    ("2024-02-10", "WN1055", "PHX", "DEN", "602", "B737", "15C", None, 45),
    ("2024-03-05", "AA892", "DEN", "DFW", "663", "B737", "21B", 198.75, None),
    ("2024-03-12", "UA634", "DFW", "IAD", "1216", "B777", "9A", 345.25, 12),
    ("2024-04-20", "B61840", "IAD", "BOS", "429", "", "11D", 179.50, 0),
    ("2024-05-15", "DL1123", "BOS", "ATL", "946", "A220", "4A", 267.00, 25),
    ("2024-05-18", "DL2967", "ATL", "MIA", "594", "B737", "", None, 8),
    ("2024-06-02", "AA1456", "MIA", "LGA", "1095", "A321", "18F", 312.80, None),
]

columns = [
    "flight_date",
    "flight_number",
    "origin",
    "destination",
    "distance",
    "aircraft",
    "seat",
    "price",
    "delay_min",
]

flights = pd.DataFrame(flights_data, columns=columns)
display(flights)
````

## Inspect before manipulating

Use `head()`, `tail()`, `shape`, `columns`, `index`, and `dtypes` to learn what you actually have. Shape lists rows before columns. String dtype labels may differ across pandas versions.

````python
display(flights.head(), flights.tail(3))
print('Shape:', flights.shape)
print('Columns:', list(flights.columns))
display(flights.dtypes, flights.index)
````

## Selecting Series, tables, labels, and positions

`df["origin"]` returns a Series; `df[["origin"]]` returns a one-column DataFrame. Prefer brackets because dot notation can collide with attributes such as `size`. `iloc` selects positions with an exclusive stop; `loc` selects labels with an inclusive label-slice stop when that slice is valid.

A default integer index makes these look similar, but labels and positions are different concepts.

````python
print(flights['origin'].shape, flights[['origin']].shape)
display(flights.iloc[2:5], flights.loc[2:5, ['origin', 'destination']])
by_number = flights.set_index('flight_number')
display(by_number.loc['UA1247'], by_number.reset_index().head(2))
weird_df = pd.DataFrame({'id':[1,2,3,4], 'size':[11,12,13,14]})
print('Total cells:', weird_df.size)
display(weird_df['size'])
````

## Boolean masks and membership

A column comparison produces one Boolean per row. Use the mask inside brackets or as the row selector in `loc`. `isin()` checks membership in a collection. Filtering retains the original row labels.

````python
mask = flights['origin'] == 'BWI'
display(mask, flights[mask])
east_coast = ['BWI','BOS','LGA','IAD']
display(flights[flights['origin'].isin(east_coast)])
display(flights.loc[flights['destination'] == 'LAX', ['origin','destination']])
````

## Missing values versus empty strings

`isna()` detects pandas missing values; `notna()` detects recorded values. An empty string is stored text, not a missing value. Here two seats are empty strings and two prices are missing. Do not compare missing values with `== NaN`.

````python
print('Missing seats:', flights['seat'].isna().sum())
print('Empty seats:', flights['seat'].eq('').sum())
display(flights[flights['seat'] == ''], flights[flights['price'].isna()])
````

## Combining conditions and query

Use `&` for elementwise AND, `|` for OR, and `~` for NOT; parenthesize each condition. Python `and`/`or` are not the operators for combining Series masks. `query()` takes an expression string and is not a SQL query.

````python
display(flights[(flights['price'] > 200) & flights['price'].notna()])
display(flights[(flights['origin'] == 'BWI') | (flights['destination'] == 'BOS')])
display(flights.query("price > 300 and origin == 'ORD'"))
````

## Types, vectorization, and assignment

Write arithmetic once to operate across a Series. But strings repeat when multiplied: `"651" * 2` is `"651651"`. Previewing `.astype(int)` returns a converted Series without modifying the original column. Assign it back to persist the conversion. A rerun of an earlier cell can then produce a different answer because the DataFrame has changed.

````python
print('Before:', (flights['distance'] * 2).iloc[0])
converted = flights['distance'].astype(int)
print('Preview:', converted.iloc[0] * 2)
print('Still unchanged:', (flights['distance'] * 2).iloc[0])
flights['distance'] = converted
flights['round_trip_distance'] = flights['distance'] * 2
assert flights['round_trip_distance'].iloc[0] == 1302
display(flights[['distance','round_trip_distance']], flights['price'] * 2)
````

## New columns and string methods

Create routes by concatenating origin and destination. Numeric division produces missing results where prices are missing. `Series[:2]` slices Series rows; `Series.str[:2]` slices every string. `.str.upper()` and `.str.contains()` also act on each string.

````python
flights['route'] = flights['origin'] + '-' + flights['destination']
flights['miles_per_dollar'] = flights['distance'] / flights['price']
flights['airline_code'] = flights['flight_number'].str[:2]
display(flights['flight_number'][:2], flights['flight_number'].str[:2])
display(flights['aircraft'].str.upper(), flights['aircraft'].str.contains('B7'))
display(flights[['route','miles_per_dollar','airline_code']])
````

## Functions, lambda, and apply

`Series.apply()` passes each value to a function. A lambda is a short unnamed function. `DataFrame.apply()` passes columns by default (`axis=0`); `axis=1` passes rows. Prefer direct column arithmetic when it expresses the same operation clearly. The lecture uses the approximate factor 1.6 for miles-to-kilometers.

````python
def tokm(x):
    return x * 1.6

display(flights['distance'].apply(tokm))
flights['flight_type'] = flights['distance'].apply(lambda x: 'Short' if x < 500 else 'Medium' if x < 1000 else 'Long')
display(flights[['origin','destination']].apply('-'.join, axis=1))
try:
    flights.apply(lambda row: row['flight_number'])
except KeyError as error:
    print('Expected error when columns are passed:', error)
display(flights.apply(lambda row: f"Flight {row['flight_number']} from {row['origin']} to {row['destination']}", axis=1))
````

## A multi-condition row function

The lecture’s function labels missing delays as “On time.” This demonstrates function syntax but introduces an assumption: unknown does not necessarily mean zero. The supplemental version below preserves “Unknown” so you can compare interpretations.

````python
def flight_experience(row):
    if pd.isna(row['delay_min']) or row['delay_min'] == 0:
        status = 'On time'
    elif row['delay_min'] < 30:
        status = 'Minor delay'
    else:
        status = 'Major delay'
    return f"{row['flight_number'][:2]} flight: {status}"

flights['experience'] = flights.apply(flight_experience, axis=1)
flights['delay_status_preserved'] = flights['delay_min'].apply(lambda x: 'Unknown' if pd.isna(x) else 'On time' if x == 0 else 'Minor delay' if x < 30 else 'Major delay')
display(flights[['flight_number','delay_min','experience','delay_status_preserved']])
````

## Dates, sorting, and method chains

Convert date strings using `pd.to_datetime()` and extract fields with `.dt`. Monday is day 0 and Sunday day 6. `sort_values()` orders rows; `nlargest()` and `nsmallest()` select numeric extremes. A chain should be read one operation at a time; sorting does not modify the original unless saved.

````python
dates = pd.to_datetime(flights['flight_date'], format='%Y-%m-%d')
flights['weekday'] = dates.dt.day_name()
flights['weekend'] = dates.dt.dayofweek >= 5
display(flights[['flight_date','weekday','weekend']])
display(flights.nlargest(3,'distance')[['flight_number','distance']], flights.nsmallest(3,'distance')[['flight_number','distance']])
expensive = (flights.loc[flights['price'].notna(), ['flight_number','origin','price']]
             .sort_values('price', ascending=False).head(3))
display(expensive)
````
