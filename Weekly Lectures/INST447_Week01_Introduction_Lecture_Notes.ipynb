## The messy middle

Data science starts with a question, not a library import. The lifecycle connects question, data, analysis, and explanation or decision. INST447 focuses on locating, acquiring, inspecting, cleaning, transforming, combining, and validating data. Code can run successfully while answering the wrong question.

At each stage ask: What do I intend to measure? What does one record represent? Is this source appropriate? Which assumptions affect the result?

## A pipeline is a sequence of decisions

Define a question; locate and acquire data; inspect and clean it; transform and combine it; analyze; communicate. Preserve raw inputs and make derived outputs reproducible. If an input changes, identify which downstream steps need rerunning.

Supplemental miniature example: unknown delays are excluded explicitly rather than silently treated as zero.

```python
raw_delays = (15, None, 0, 45)
known_delays = [d for d in raw_delays if d is not None]
mean_delay = sum(known_delays) / len(known_delays)
assert raw_delays == (15, None, 0, 45)
assert len(known_delays) == 3
print({'records': len(raw_delays), 'known_delays': len(known_delays), 'mean_known_delay': mean_delay})
```

## Notebook state and repeatable functions

Notebooks combine narrative, exploratory code, and immediate output. Scripts and functions help repeat stable logic. Neither format guarantees correctness. Hidden state occurs when later calculations depend on cells run out of visible order. Restart and run top to bottom.

Supplemental example: put a repeatable calculation into a function with explicit inputs and checks.

```python
def average_known(values):
    known = [value for value in values if value is not None]
    if not known:
        return None
    return sum(known) / len(known)

assert average_known([15, None, 0, 45]) == 20
assert average_known([None]) is None
print(average_known(raw_delays))
```
