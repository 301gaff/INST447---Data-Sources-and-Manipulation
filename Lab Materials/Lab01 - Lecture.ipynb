## Pandas Bascis

```python
# import the pandas library
import pandas as pd
```

```python
url = (
    "https://raw.githubusercontent.com/"
    "allisonhorst/palmerpenguins/main/inst/extdata/penguins.csv"
)

df = pd.read_csv(url)
```

```python
# Display the first five rows
df.head()
```

```python
# Examine the structure of the table
df.shape
```

```python
# View column names
df.columns
```

```python
# Examine column data types and missing values
df.info()
```

```python
# Generate descriptive statistics
df.describe()
```

```python
# Select one column
df["body_mass_g"]
```

```python
# Select multiple columns
df[["body_mass_g", "sex"]]
```

```python
# Filter rows
df[df["body_mass_g"] > 3500]
```

```python
# Count missing values in each column
df.isna().sum()
```
